```python
import requests
import pandas as pd
import numpy as np
import xmltodict

import pprint
```

```python
url_trafico="https://datos.madrid.es/egob/catalogo/202584-0-aparcamientos-residentes.xml"
r=requests.get(url_trafico)
r
```

```python
contenido=r.content
```

```python
my_dict = xmltodict.parse(contenido)
my_dict
```

```python
for x in range(len(my_dict["Contenidos"]["contenido"])):
    print(my_dict["Contenidos"]["contenido"][x]["atributos"]["atributo"])
```
