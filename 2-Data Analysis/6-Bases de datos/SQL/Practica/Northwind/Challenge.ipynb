Pista: Utiliza gráficos de barra y de dispersión para comparar el desempeño de los empleados y identificar áreas de mejora.

```python
engine = create_engine("sqlite:///data/northwind.db", echo=False)
connection = engine.connect() #para sacar en dataframe
```

Todos los empleados venden mas o menos lo mismo. En cuant a ventas y a dinero generado.

```python
query="""
SELECT o.employeeid,FirstName, LastName, count(O.orderid) as ventas , sum(unitprice*quantity-discount) as dinero_generado
from employees as e
inner join orders as o on e.employeeid=o.employeeid
inner join "Order Details" as od on o.OrderID=od.OrderID
group by o.employeeid
order by ventas desc
"""
df5=pd.read_sql(query, connection)
df5
```

```python
trace1 = go.Bar(x = df5["EmployeeID"],
               y = df5['ventas'],
               name = 'Ventas de cada empleado',
               marker = dict(color = 'rgba(255, 174, 255, 0.5)',
                            line = dict(color='rgb(0,0,0)', width = 1.5)),
               text = df5['LastName'])


data = trace1

layout = go.Layout(barmode = "group")#(puedes meter overlay) #laben

fig = go.Figure(data = data, layout = layout)

iplot(fig)
```

```python
trace1 = go.Bar(x = df5["EmployeeID"],
               y = df5['dinero_generado'],
               name = 'Ventas de cada empleado',
               marker = dict(color = 'rgba(255, 174, 255, 0.5)',
                            line = dict(color='rgb(0,0,0)', width = 1.5)),
               text = df5['LastName'])


data = trace1

layout = go.Layout(barmode = "group")#(puedes meter overlay) #laben

fig = go.Figure(data = data, layout = layout)

iplot(fig)
```
