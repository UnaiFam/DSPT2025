## Ejercicio 15
Crea un array cuadrado de 2D, con las dimensiones que quieras, pero que todo el borde esté compuesto por 1s, y el interior por 0s

````python
def cubo(lado):
    #hago la tapa, en forma de lista para luego pasarlo a array. str para copiar el numero al multiplicar
    lista_tapa=list(("1"*lado))
    
    #hago el grueso de la mima forma pero ahora el centro es 0
    lista_grueso=list("1"+("0"*(lado-2))+"1")
    #print(lista_tapa)
    #print(lista_grueso)
    #ahora ahgo la lista final solo con las tapas del cuadrado, luego le metere el grueso en medio
    lista_cubo=[lista_tapa, lista_tapa]
    # ahora meto el grueso en el medio (posiocion 1) tantas copias como me hagn falta (resto la tapas)
    for i in range(lado-2):
        lista_cubo.insert(1, lista_grueso)
    #ahora paso la lista a array y lo paso a int
    my_array=np.array(lista_cubo)
    #no estoy seguro de como pasar a int el str
    my_arrayint=np.astype(my_array, int)
    return my_arrayint
````
