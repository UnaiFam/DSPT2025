### Step 3. Veamos los tipos de datos. ¿Podríamos obtener el producto de mayor precio?

```python
data.info()
```

```python
# gnereo floats de item
data["itempricefloat"]=(data["item_price"].str.replace("$", "").astype(float))#str te da acceso al objeto y la serie
#genero de cada iimpes
data["individual_price"]=data["itempricefloat"]/data["quantity"]
data["item_name"][(data["itempricefloat"].idxmax())]
```

```python
valor_maximo=data["itempricefloat"].max()
condicion=(data["itempricefloat"]==valor_maximo)

data[condicion]
```

### Step 4. ¿Qué productos cuestan más de $10?

```python
data["item_name"][data["itempricefloat"]>10].unique()
```

```python
condicion=(data["itempricefloat"]>10)
data[condicion]
len(data[condicion])
```

### Step 4.1: ¿Y cuántos pedidos se han hecho con un producto de más de 10$? ¿Es lo mismo?

```python
condicion=data["itempricefloat"]>=10
data[condicion]["order_id"]
```

### Step 4.2: ¿Y cuántos pedidos se han hecho de más de 10$? ¿Es lo mismo?

```python
#hacerlo con group by
data_id10=data.groupby("order_id")["itempricefloat"].sum()
data_id10
```

### Step 4.3: ¿Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? ¿Es lo mismo?

```python
condicion=data["individual_price"]>=10
data[condicion]["order_id"]
```

### Step 5. ¿Qué precio tiene cada producto en distintos pedidos? ¿Hay productos con varios precios?

```python
data_name=data.groupby("item_name")["individual_price"].unique()
print(data_name, "Si")
```

### Step 7. ¿Cuantas veces se ha pedido los productos más caros?
