## Ejercicio 2
En este ejercicio vamos a implementar un calculador de precios de casas muy sencillo. Tenemos las siguientes variables:

> superficie
>
> distrito

Implementa la siguiente casuística:

1. Si el distrito es "Moncloa" o "Centro", y además la superficie es superior a 100 metros cuadrados, el precio de la casa es de 1000
2. Si el distrito es "Salamanca", y además la superficie de la casa es al menos de 150 metros, el precio de la casa es de 1500
3. Si el distrito no es "Retiro" y la superficie está entre 60 y 80 metros, el precio es de 600
4. En cualquier otro caso, el precio será de 0

```python
# AQUÍ VA VUESTRO CÓDIGO
distrito=input()
superficie=float(input())
if distrito=="Moncloa" or "Centro" and superficie>100:
    precio=1000
elif distrito=="Salamanca" and superficie>=150:
    precio=1500
elif distrito=="Retiro" and 60<superficie<80:
    precio=600   
else:
    precio=0
print(precio)
```

## Ejercicio 3
En este ejercicio vamos a realizar un programa muy parecido. Para este caso queremos que se cumplan las siguientes condiciones:

1. Primero se compruebe si el distrito es "Retiro". Si es asá, que imprima "Distrito Retiro", y si no, "Otro distrito"
2. Si el distrito es "Retiro", comprueba si la superficie es mayor de 100 metros cuadrados. En tal caso, que imprima un precio de 1000, y si no, de 500.

Prueba a usar `ifs` anidados

```python
# AQUÍ VA VUESTRO CÓDIGO
distrito=input()
superficie=float(input())
if distrito=="Retiro":
    print("Distrito Retiro")
    if superficie>=100:
        precio=1000
    else:
        precio=500
else:
    precio="Otro distrito"
print(precio)
```

## Ejercicio 6
Dada la siguiente lista:

> list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

Implementa un programa que los recorra e imprima por pantalla todos los divisibles por 5. Si nos encontramos con alguno que sea mayor que 150, detener el bucle.

```python
# AQUÍ VA VUESTRO CÓDIGO
list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

for i in list1:
    if i>150:
        break
    elif i%5==0:
        print(i)
```

NOTA: NO hay lineas en blanco entre una línea y otra.

3. Dentro del bucle, obtener el valor del usuario y comprobar si es ese el numero a adivinar. Si no, actualizar las vidas.
4. Si acertamos, salimos del bucle e imprimimos por pantalla "You win". Y si perdemos también salimos del bucle, pero en este caso imprimimos por pantalla "You lose".

TIP: te puede resultar útil usar la sentencia `else` cuando acabe el bucle `while`. Lo que haya dentro de ese `else` se ejecutará una vez acabe la ejecución del `while`. Lo podrás usar para cuando pierdas.

```python
import random
```

```python
# AQUÍ VA VUESTRO CÓDIGO
#random.randint(0,5)
#help(random)
print(random.randrange(1,6))
```

```python
vidas=2
while vidas>0:
    respuesta=int(input("Adivina el numero que estoy pensando"))
    numero_pensado=random.randrange(1,6)
    if numero_pensado==respuesta:
        print("You win")
        break
    if respuesta>5 or respuesta<0 :
        print("respuesta no valida")
    elif numero_pensado != respuesta:
        print ("No es ese")
        vidas-=1
        print(f"Te quedan {vidas} vidas")
if vidas==0:
    print("You lose")
```
