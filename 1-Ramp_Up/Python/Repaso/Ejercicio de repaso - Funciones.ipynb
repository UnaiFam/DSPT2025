```python
def palabros(palabra1, palabra2):
    list1=[]
    lista_solucion=[]
    for i in palabra1.lower():
        list1.append(i)
    set1=set(list1)
    #esto se puede evidtas con sets
    #&compara dos set o .intersection() solo faltaria pasar a numero
    for letra in palabra2.lower():
        if letra in set1:
            lista_solucion.append(letra)

    return len(lista_solucion)
```

### Ejercicio 2
Un número primo es aquel número que solo es divisible por 1 y por si mismo (2,3,5,7,11,13,...). Define una función que pida al usuario un número natural mayor que 2 y devuelva si se trata de un número primo o no.

```python
def primo(num):
    if num <=2 or type(num)!=int:
        return "no es valido"
    elif type(num)==int:
        for i in range(2, num):
            if num%i==0:
                return "No es primo"

        return "Es primo"
```

```python
primo(2525)
```

### Ejercicio 5 
Definir una función que reciba una cadena de caracteres y devuelva un diccionario cuyas claves sean las cinco vocales y sus valores sean sus frecuencias dentro de la cadena respectivamente.

```python
def diccionarion (caracteres):
    diccionario_salida ={}
    vocales="aeiou"
    for i in caracteres:
        if i in vocales:
            diccionario_salida[i]=caracteres.count(i)
    return diccionario_salida
```

### Ejercicio 7
Desarrollar una función que reciba una lista de números y devuelva una lista de listas (lista = [$\text{sublista}_{1}$,$\text{sublista}_{2}$,$\text{sublista}_{3}$]), en la que cada sublista contendrá determinados números.

* sublista1: los números que tengan de 1-6 cifras.
* sublista2: los números que tengan de 7-12 cifras.
* sublista3: los números que tengan m´as de 12 cifras.

**Nota**: cuidado cuando tengamos un número negativo o con decimales.


```python
def lista_numeral (lista):
    lista16=[]
    lista712=[]
    lista13=[]
    
    for i in lista:
        if len(str(i))<6:
            lista16.append(i)
        if 7<len(str(i))<12:
            lista712.append(i)
        if len(str(i))>12:
            lista13.append(i) 

    return [lista16, lista712, lista13]
```
